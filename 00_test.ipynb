```python
# Student grade A - above 95 marks, B - above 85 marks, C - above 75 marks, D - above 65 marks, E - above 55 marks, F - below 55 marks
# if elif else format 
marks = int(input('Enter your marks: '))
if marks > 95:
    print('Grade A')
elif marks > 85:
    print('Grade B')
elif marks > 75:
    print('Grade C')
elif marks > 65:
    print('Grade D')
elif marks > 55:
    print('Grade E')
else:
    print('Grade F')
```
